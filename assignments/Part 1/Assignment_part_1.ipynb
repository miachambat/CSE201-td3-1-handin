## Journey length

The Suisse Impossible Mission Force suspects that the rogue agent is an anomaly within his declared employment category, particularly in terms of the number of days he spent at destinations compared to his peers.

To identify such deviations, we can employ boxplots for each employment category, allowing us to visually identify outliers in the data.

First, let's construct two tables: one for the dates of arrival and another for the dates of departure. Utilize the existing dataset to construct these tables based on the 'date', 'destination' and 'departure' columns (again note that 'date' field represents both departure and arrival date). Only consider arrival and departure flights whose date is between 1 September 2019 and 31 October 2020 (inclusive).

```python
# For this step we will need the information not only about the fights to the US,
# but also about the flights taken from the US

# Your code here

us_travels_toandfrom = merged_data[
    ((merged_data['destination'] == 'US') | (merged_data['departure'] == 'US')) &
    (merged_data['date'] >= start_date) &
    (merged_data['date'] <= end_date)
].copy()

cols_to_keep = ['UserID', 'date', 'destination', 'departure', 'employment']
all_US_data = us_travels_toandfrom[cols_to_keep]
print(all_US_data.head())
print(all_US_data.shape)

# Two separate tables

arrivals = us_travels_toandfrom[us_travels_toandfrom['destination'] == 'US'][['UserID', 'date', 'employment']].copy()
arrivals = arrivals.rename(columns={'date': 'arrival_date'})

departures = us_travels_toandfrom[us_travels_toandfrom['departure'] == 'US'][['UserID', 'date', 'employment']].copy()
departures = departures.rename(columns={'date': 'departure_date'})

print(f"Arrivals found: {len(arrivals)}")
print(f"Departures found: {len(departures)}")

print("\nFirst few rows of Arrivals:")
display(arrivals.head())

print("\nFirst few rows of Departures:")
display(departures.head())

# Create sets of unique IDs
ids_arrivals = set(arrivals['UserID'])
ids_departures = set(departures['UserID'])

# Calculate the overlaps and differences
common_ids = ids_arrivals.intersection(ids_departures)
only_arrivals = ids_arrivals - ids_departures
only_departures = ids_departures - ids_arrivals
total_unique = ids_arrivals.union(ids_departures)

# Print the report
print(f"Total Unique Users Overall: {len(total_unique)}")
print(f"---")
print(f"Users in both (Inner Join match): {len(common_ids)}")
print(f"Users in Arrivals but NOT Departures: {len(only_arrivals)}")
print(f"Users in Departures but NOT Arrivals: {len(only_departures)}")
```

We can keep on both of these tables the user ID, employment, and departure date / arrival date.

```python
# Your code here
# SEE ABOVE
```

We can now join the two tables on user ID.

*Note that each suspect has at most one arrival flight to the US and one departure flight from the US during the investigation period. Therefore, when merging the arrival and departure tables, you should expect only one matched row per user ID.*

```python
# Your code here
stays = pd.merge(arrivals, departures, on=['UserID', 'employment'], how='outer', indicator=True)
# This adds a column '_merge' with values: 'both', 'left_only', or 'right_only'
print(stays['_merge'].value_counts())
print(stays.head())
```

We can now create a new feature called journey_length representing the difference between departure date and arrival date.

*Hint: Ensure that journey_length is of type numeric as the box plot in the next question requires a numeric type. To convert from datetime to number of days, you can use `dt.days`*

```python
# Your code here
stays['journey_length'] = stays['departure_date'] - stays['arrival_date']
stays['journey_length_days'] = stays['journey_length'].dt.days
stays[['UserID', 'departure_date', 'arrival_date', 'journey_length_days', 'employment']].head()
```

*Sidenote: if you get negative journey length, what would be your explanation to that?*

We can now create the boxplots where the x-axis represents each profession and the y-axis corresponds to the journey length.

Based on these visualizations, we can look for the profession containing the most extreme outlier, as our intelligence suggests the suspect acts as an anomaly compared to their peers in that category.

```python
# Your code here
import seaborn as sns
plt.figure(figsize=(14, 7))
sns.boxplot(data=stays, x='employment', y='journey_length_days', palette='Set2')

# Adding labels and formatting for clarity
plt.title('Journey length distribution by profession', fontsize=15)
plt.xlabel('Profession', fontsize=12)
plt.ylabel('Journey length (days)', fontsize=12)
plt.xticks(rotation=45) # Rotates labels to prevent overlap
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.savefig('journey_length_boxplot.png')

# Group by Employment and get the max journey_length for each
outliers_by_employment = stays.groupby('employment')['journey_length_days'].max().sort_values(ascending=False)

print("\n--- Longest Stay per Employment Category ---")
print(outliers_by_employment)

comparison_df = stays.groupby('employment')['journey_length_days'].agg(['max', 'mean', 'count'])
comparison_df['deviation_from_avg'] = comparison_df['max'] - comparison_df['mean']
comparison_df = comparison_df.sort_values(by='max', ascending=False)
print(comparison_df)

comparison_min_df = stays.groupby('employment')['journey_length_days'].agg(['min', 'mean', 'count'])
comparison_min_df['deviation_from_avg'] = comparison_min_df['min'] - comparison_min_df['mean']
comparison_min_df = comparison_min_df.sort_values(by='min', ascending=True)
print("\n--- Shortest Stay Comparison by Employment Category ---")
print(comparison_min_df)
```
